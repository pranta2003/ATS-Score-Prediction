```python

#try foe split the resume and job discription


def split_resume_job(text):

    if " SEP " in text:

        resume, job = text.split(" SEP ", 1)

    else:

        resume = text
        job = ""

    return pd.Series([resume, job])
```

```python
train[["resume","job_description"]] = train["text"].apply(split_resume_job)

valid[["resume","job_description"]] = valid["text"].apply(split_resume_job)
```

```python
#Cleaning data

import re

def clean_text(text):

    text = str(text)

    text = text.lower()

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"www\S+", "", text)

    text = re.sub(r"<.*?>", "", text)

    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
import re

def preprocess_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove HTML
    text = re.sub(r'<.*?>', ' ', text)

    # Add space between letters and numbers
    text = re.sub(r'([a-zA-Z])(\d)', r'\1 \2', text)
    text = re.sub(r'(\d)([a-zA-Z])', r'\1 \2', text)

    # Add space before capital letters in joined words
    text = re.sub(r'([a-z])([A-Z])', r'\1 \2', text)

    # Keep useful symbols
    text = re.sub(r'[^a-z0-9+#.\-\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
resume_embeddings = model.encode(
    train["resume"].tolist(),
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
job_embeddings = model.encode(
    train["job_description"].tolist(),
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
print("Resume Shape:", resume_embeddings.shape)
print("Job Shape:", job_embeddings.shape)
```

```python
import numpy as np

X = np.concatenate(
    (resume_embeddings, job_embeddings),
    axis=1
)

y = train["ats_score"].values
```

```python
print("X Shape :", X.shape)
print("y Shape :", y.shape)
```

```python
#Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
#K-Selection
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error

best_k = 0
best_mae = float("inf")

print("Validation Results")
print("-"*30)

for k in [1,3,5,7,9]:

    knn = KNeighborsRegressor(n_neighbors=k)

    knn.fit(X_train, y_train)

    pred = knn.predict(X_test)

    mae = mean_absolute_error(y_test, pred)

    print(f"k = {k}   MAE = {mae:.4f}")

    if mae < best_mae:
        best_mae = mae
        best_k = k

print("\nBest k =", best_k)
print("Best MAE =", best_mae)
```

```python
from sklearn.model_selection import KFold
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from statistics import mean
import numpy as np

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=32
)

mae_scores = []
rmse_scores = []
r2_scores = []

fold = 1

for train_index, test_index in kfold.split(X):

    X_train_cv = X[train_index]
    X_test_cv = X[test_index]

    y_train_cv = y[train_index]
    y_test_cv = y[test_index]

    model = KNeighborsRegressor(
        n_neighbors=best_k
    )

    model.fit(X_train_cv, y_train_cv)

    pred = model.predict(X_test_cv)

    mae = mean_absolute_error(y_test_cv, pred)
    rmse = np.sqrt(mean_squared_error(y_test_cv, pred))
    r2 = r2_score(y_test_cv, pred)

    mae_scores.append(mae)
    rmse_scores.append(rmse)
    r2_scores.append(r2)

    print(f"\nFold {fold}")
    print("MAE :", mae)
    print("RMSE :", rmse)
    print("R² :", r2)

    fold += 1

print("\n========== Cross Validation Results ==========")
print("Average MAE :", mean(mae_scores))
print("Average RMSE:", mean(rmse_scores))
print("Average R² :", mean(r2_scores))
```

```python
#Stacking applying as ensembel learning 


from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor



estimators = [

    ("lr", LinearRegression()),

    ("knn", KNeighborsRegressor(
        n_neighbors=best_k
    )),

    ("rf", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    )),

    ("xgb", XGBRegressor(
        n_estimators=200,
        learning_rate=0.1,
        random_state=42
    ))

]


stack = StackingRegressor(
    estimators=estimators,
    final_estimator=LinearRegression(),
    cv=5,
    n_jobs=-1
)


stack.fit(X_train, y_train)


stack_pred = stack.predict(X_test)

print("Stacking Regressor")

print("MAE :", mean_absolute_error(y_test, stack_pred))
print("RMSE :", np.sqrt(mean_squared_error(y_test, stack_pred)))
print("R² :", r2_score(y_test, stack_pred))
```
