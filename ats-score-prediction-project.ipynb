```python
#Cleaning data

import re

def clean_text(text):

    text = str(text)

    text = text.lower()

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"www\S+", "", text)

    text = re.sub(r"<.*?>", "", text)

    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
import re

def preprocess_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove HTML
    text = re.sub(r'<.*?>', ' ', text)

    # Add space between letters and numbers
    text = re.sub(r'([a-zA-Z])(\d)', r'\1 \2', text)
    text = re.sub(r'(\d)([a-zA-Z])', r'\1 \2', text)

    # Add space before capital letters in joined words
    text = re.sub(r'([a-z])([A-Z])', r'\1 \2', text)

    # Keep useful symbols
    text = re.sub(r'[^a-z0-9+#.\-\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()
```

```python
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score
from statistics import mean

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=32
)

# Store R² scores
lr_train_r2 = []
lr_val_r2 = []

knn_train_r2 = []
knn_val_r2 = []

fold = 1

for train_index, val_index in kfold.split(X):

    X_train_cv = X[train_index]
    X_val_cv = X[val_index]

    y_train_cv = y[train_index]
    y_val_cv = y[val_index]

    # =========================
    # Linear Regression
    # =========================

    lr = LinearRegression()

    lr.fit(X_train_cv, y_train_cv)

    lr_train_pred = lr.predict(X_train_cv)
    lr_val_pred = lr.predict(X_val_cv)

    lr_train_score = r2_score(y_train_cv, lr_train_pred)
    lr_val_score = r2_score(y_val_cv, lr_val_pred)

    lr_train_r2.append(lr_train_score)
    lr_val_r2.append(lr_val_score)


    # =========================
    # KNN Regression
    # =========================

    knn = KNeighborsRegressor(
        n_neighbors=best_k
    )

    knn.fit(X_train_cv, y_train_cv)

    knn_train_pred = knn.predict(X_train_cv)
    knn_val_pred = knn.predict(X_val_cv)

    knn_train_score = r2_score(y_train_cv, knn_train_pred)
    knn_val_score = r2_score(y_val_cv, knn_val_pred)

    knn_train_r2.append(knn_train_score)
    knn_val_r2.append(knn_val_score)


    print(f"\n========== Fold {fold} ==========")

    print("Linear Regression")
    print("Training R²   :", lr_train_score)
    print("Validation R² :", lr_val_score)

    print("\nKNN Regressor")
    print("Training R²   :", knn_train_score)
    print("Validation R² :", knn_val_score)

    fold += 1


# =========================
# Average Results
# =========================

print("\n\n======================================")
print("       AVERAGE TRAIN vs VALIDATION")
print("======================================")

print("\nLinear Regression")
print("Average Training R²   :", mean(lr_train_r2))
print("Average Validation R² :", mean(lr_val_r2))
print("R² Gap                :", mean(lr_train_r2) - mean(lr_val_r2))


print("\nKNN Regressor")
print("Average Training R²   :", mean(knn_train_r2))
print("Average Validation R² :", mean(knn_val_r2))
print("R² Gap                :", mean(knn_train_r2) - mean(knn_val_r2))
```

```python
#allready done bagging on knn and linear but still using random forest and xgboost as model but not for any bagging or  boosting .

from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score
from statistics import mean

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=32
)

models = {
    "Linear Regression": LinearRegression(),

    "KNN": KNeighborsRegressor(
        n_neighbors=best_k
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBRegressor(
        n_estimators=200,
        learning_rate=0.1,
        random_state=42,
        n_jobs=-1
    )
}

results = {}

for name, model in models.items():

    train_r2_scores = []
    val_r2_scores = []

    for train_index, val_index in kfold.split(X):

        X_train_cv = X[train_index]
        X_val_cv = X[val_index]

        y_train_cv = y[train_index]
        y_val_cv = y[val_index]

        model.fit(X_train_cv, y_train_cv)

        train_pred = model.predict(X_train_cv)
        val_pred = model.predict(X_val_cv)

        train_r2_scores.append(
            r2_score(y_train_cv, train_pred)
        )

        val_r2_scores.append(
            r2_score(y_val_cv, val_pred)
        )

    avg_train_r2 = mean(train_r2_scores)
    avg_val_r2 = mean(val_r2_scores)

    results[name] = (avg_train_r2, avg_val_r2)

    print("\n", name)
    print("Average Training R²   :", avg_train_r2)
    print("Average Validation R² :", avg_val_r2)
    print("Train-Validation Gap  :", avg_train_r2 - avg_val_r2)
```

```python
#using cleaning function 

import re
from bs4 import BeautifulSoup

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r"http\S+|www\S+", " ", text)

    text = BeautifulSoup(text, "html.parser").get_text(" ")

    text = re.sub(r"\S+@\S+", " ", text)

    text = re.sub(r"\+?\d[\d\s().-]{7,}\d", " ", text)

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
#ATS prediction creating function 
def predict_ats(resume, job_description):

    resume_clean = clean_text(resume)
    job_clean = clean_text(job_description)

    resume_embedding = sbert_model.encode(resume_clean)
    job_embedding = sbert_model.encode(job_clean)

    combined_features = np.concatenate([
        resume_embedding,
        job_embedding
    ])

    combined_features = combined_features.reshape(1, -1)

    lr_prediction = bag_lr.predict(combined_features)[0]
    knn_prediction = bag_knn.predict(combined_features)[0]

    ats_score = (lr_prediction + knn_prediction) / 2

    return float(ats_score)
```
