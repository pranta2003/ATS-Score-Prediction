```python

#try foe split the resume and job discription


def split_resume_job(text):

    if " SEP " in text:

        resume, job = text.split(" SEP ", 1)

    else:

        resume = text
        job = ""

    return pd.Series([resume, job])
```

```python
#Cleaning data

import re

def clean_text(text):

    text = str(text)

    text = text.lower()

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"www\S+", "", text)

    text = re.sub(r"<.*?>", "", text)

    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
import re

def preprocess_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove HTML
    text = re.sub(r'<.*?>', ' ', text)

    # Add space between letters and numbers
    text = re.sub(r'([a-zA-Z])(\d)', r'\1 \2', text)
    text = re.sub(r'(\d)([a-zA-Z])', r'\1 \2', text)

    # Add space before capital letters in joined words
    text = re.sub(r'([a-z])([A-Z])', r'\1 \2', text)

    # Keep useful symbols
    text = re.sub(r'[^a-z0-9+#.\-\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()
```

```python
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, BaggingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score
from statistics import mean

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=32
)

# Store validation R²
results = {
    "Linear Regression": [],
    "KNN": [],
    "Random Forest": [],
    "XGBoost": [],
    "Bagging LR + KNN": []
}

fold = 1

for train_index, val_index in kfold.split(X):

    X_train_cv = X[train_index]
    X_val_cv = X[val_index]

    y_train_cv = y[train_index]
    y_val_cv = y[val_index]

    # =========================
    # Linear Regression
    # =========================

    lr = LinearRegression()

    lr.fit(X_train_cv, y_train_cv)

    lr_val_pred = lr.predict(X_val_cv)

    lr_val_r2 = r2_score(
        y_val_cv,
        lr_val_pred
    )

    results["Linear Regression"].append(lr_val_r2)


    # =========================
    # KNN
    # =========================

    knn = KNeighborsRegressor(
        n_neighbors=best_k
    )

    knn.fit(X_train_cv, y_train_cv)

    knn_val_pred = knn.predict(X_val_cv)

    knn_val_r2 = r2_score(
        y_val_cv,
        knn_val_pred
    )

    results["KNN"].append(knn_val_r2)


    # =========================
    # Random Forest
    # =========================

    rf = RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )

    rf.fit(X_train_cv, y_train_cv)

    rf_val_pred = rf.predict(X_val_cv)

    rf_val_r2 = r2_score(
        y_val_cv,
        rf_val_pred
    )

    results["Random Forest"].append(rf_val_r2)


    # =========================
    # XGBoost
    # =========================

    xgb = XGBRegressor(
        n_estimators=100,
        learning_rate=0.1,
        random_state=42,
        n_jobs=-1
    )

    xgb.fit(X_train_cv, y_train_cv)

    xgb_val_pred = xgb.predict(X_val_cv)

    xgb_val_r2 = r2_score(
        y_val_cv,
        xgb_val_pred
    )

    results["XGBoost"].append(xgb_val_r2)


    # =========================
    # Bagging Linear Regression
    # =========================

    bag_lr = BaggingRegressor(
        estimator=LinearRegression(),
        n_estimators=10,
        random_state=42,
        n_jobs=-1
    )

    bag_lr.fit(X_train_cv, y_train_cv)

    bag_lr_pred = bag_lr.predict(X_val_cv)


    # =========================
    # Bagging KNN
    # =========================

    bag_knn = BaggingRegressor(
        estimator=KNeighborsRegressor(
            n_neighbors=best_k
        ),
        n_estimators=10,
        random_state=42,
        n_jobs=-1
    )

    bag_knn.fit(X_train_cv, y_train_cv)

    bag_knn_pred = bag_knn.predict(X_val_cv)


    # Combine Bagging predictions
    bag_pred = (
        bag_lr_pred +
        bag_knn_pred
    ) / 2

    bag_r2 = r2_score(
        y_val_cv,
        bag_pred
    )

    results["Bagging LR + KNN"].append(bag_r2)


    # =========================
    # Print Fold Results
    # =========================

    print(f"\n========== Fold {fold} ==========")

    print("Linear Regression R² :", lr_val_r2)
    print("KNN R²               :", knn_val_r2)
    print("Random Forest R²     :", rf_val_r2)
    print("XGBoost R²           :", xgb_val_r2)
    print("Bagging LR + KNN R²  :", bag_r2)

    fold += 1


# =========================
# Average Validation R²
# =========================

print("\n\n======================================")
print("       5-FOLD VALIDATION RESULTS")
print("======================================")

for name, scores in results.items():

    print(f"\n{name}")
    print("Average Validation R² :", mean(scores))
```

```python
#allready done bagging on knn and linear but still using random forest and xgboost as model but not for any bagging or  boosting .

from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score
from statistics import mean

kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=32
)

models = {
    "Linear Regression": LinearRegression(),

    "KNN": KNeighborsRegressor(
        n_neighbors=best_k
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBRegressor(
        n_estimators=200,
        learning_rate=0.1,
        random_state=42,
        n_jobs=-1
    )
}

results = {}

for name, model in models.items():

    train_r2_scores = []
    val_r2_scores = []

    for train_index, val_index in kfold.split(X):

        X_train_cv = X[train_index]
        X_val_cv = X[val_index]

        y_train_cv = y[train_index]
        y_val_cv = y[val_index]

        model.fit(X_train_cv, y_train_cv)

        train_pred = model.predict(X_train_cv)
        val_pred = model.predict(X_val_cv)

        train_r2_scores.append(
            r2_score(y_train_cv, train_pred)
        )

        val_r2_scores.append(
            r2_score(y_val_cv, val_pred)
        )

    avg_train_r2 = mean(train_r2_scores)
    avg_val_r2 = mean(val_r2_scores)

    results[name] = (avg_train_r2, avg_val_r2)

    print("\n", name)
    print("Average Training R²   :", avg_train_r2)
    print("Average Validation R² :", avg_val_r2)
    print("Train-Validation Gap  :", avg_train_r2 - avg_val_r2)
```

```python
#using cleaning function 

import re
from bs4 import BeautifulSoup

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r"http\S+|www\S+", " ", text)

    text = BeautifulSoup(text, "html.parser").get_text(" ")

    text = re.sub(r"\S+@\S+", " ", text)

    text = re.sub(r"\+?\d[\d\s().-]{7,}\d", " ", text)

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
#ATS prediction creating function 
def predict_ats(resume, job_description):

    resume_clean = clean_text(resume)
    job_clean = clean_text(job_description)

    resume_embedding = sbert_model.encode(resume_clean)
    job_embedding = sbert_model.encode(job_clean)

    combined_features = np.concatenate([
        resume_embedding,
        job_embedding
    ])

    combined_features = combined_features.reshape(1, -1)

    lr_prediction = bag_lr.predict(combined_features)[0]
    knn_prediction = bag_knn.predict(combined_features)[0]

    ats_score = (lr_prediction + knn_prediction) / 2

    return float(ats_score)
```
